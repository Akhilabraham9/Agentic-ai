### The Agent Landscape

1. Builders - n8n, ElevenLabs, OpenAI Agent Builder, CrewAI studio 
    - *(Non-technical can **build** AI Agent)*
2. Products - Claude Code, Cowork, OpenClaw, Claude Design 
    - *(Non-technical people can **use** AI Agents)*
3. Runtimes - AWS Lambda, AWS Bedrock AgentCore, Claude Managed Agents, Vertex AI Agent Engine 
    - *(Technical people can **execute** AI Agent)*
4. Framework- OpenAI Agents SDK, CrewAIm LangGraph/LangChain, Google ADK 
    - *(Technical people can **develop** AI Agents)*

### Agent Frameworks

They are basically libraries that contains utilities, Helper code that make it easier to create Agents based on LLM calls

* Orchestration 
    - chaining together LLM calls
* Tool Calling  
    - equiping an LLM with abilities including with MCP
* Constructing LLM inputs and interpreting outputs 
    - including a technique called 'Structured output'
