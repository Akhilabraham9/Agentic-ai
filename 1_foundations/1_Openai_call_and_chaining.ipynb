```python

# Next it's time to load the API keys into environment variables
# If this returns false, see the next cell!
load_dotenv(override = True)

# Get the open api key from .env
api_key = os.getenv("OPENAI_API_KEY")
```

```python
# Create an OpenAI python client for making web calls to OpenAI
openai = OpenAI()

# Create json string to pass to OPENAI method to make LLM call
# It has to be in LIST as we can send multiple instructions
messages = [{
    'role': 'user',
    'content': 'Tell me a fun fact'
}]

# Get the response from 
response = openai.chat.completions.create(
    model = "gpt-5.4-nano",
    messages = messages
)

# Response will bring a json string in which response will be found in choices[0] -> message -> content
print(response.choices[0].message.content)
```
