# Agentic system

**Workflows** are systems where LLMs and tools are orchestrated through predefined code paths

**Agents** are dynamic system where LLMs direct their own processses and tool usage, maintaining control over how they accomplish tasks.

- **Prompt chaining:** Break a task into sequential steps, where each step uses the previous step’s output <br>
    : `Input → Step 1 → Step 2 → Output`
- **Routing:** Send an input to the best-matched path or tool <br>
    : `Input → Router → Specialized path → Output`
- **Parallelization:** Run independent tasks at the same time and combine their results <br>
    : `Input → [Task A ‖ Task B] → Combine → Output`
- **Orchestrator–worker:** An orchestrator splits work among workers and combines their results <br>
    : `Request → Orchestrator → Workers → Orchestrator → Output`
- **Evaluator–optimizer:** An evaluator critiques a result so an optimizer can improve it in a feedback loop <br>
    : `Draft → Evaluator → Feedback → Optimizer → Revised draft ↺`

    Also called LLM-as-judge

### Risks of Agent Framework

1. Unpredicatable path
2. Unpredicatable output
3. Unpredicatable cost

How to resolve
1. Monitor (Observability, Evals)
    - Understand what is actually going on with LLM, have right kind of info - traces
    - eval against end-commercial goal 
    - using LLM as evaluator/ LLM as a judge.
2. Guardrails
    - Ensure your agents behave safely, consistently and within your intended boundaries
    - Check at input or outputs

### First Agent Trap 

Business owner - "I need an Agent for my business"

Developer - "OK - what business problem are we trying to solve"

Business owner - "I want a strategy Agent to give me advice for my org"

Developer - " OK right and what business problem is that solving"


*SOLUTION - BE SURE THAT YOU RE ALWAYS SOLVING A PROBLEM*

### Second Agent Trap

Anthropomorphizing :
    A common red flag: 'Agent Architectures' that assign human-like responsibilities to agent

                                                        TRADING MANAGER AGENT                               

                                MARKET RESEARCH AGENT       TRADER AGENT           RISK MANAGER AGENT       

*SOLUTION - ALWAYS START SIMPLE: Measure performance against a business objective. Add Agents because it improves performance*


### Takeaways

1. Build Agent systems to solve problems in a measurable way
2. Decide your Agent Architecture based on performance
3. LLMs generate plausible outputs, not necessarily accurate outputs. Align the output with business results? That's your job
